```python
import matplotlib.pyplot as plt
import numpy as np
from skimage import data, exposure


img = data.moon()

p3, p80 = np.percentile(img, (3, 80))

img_rescale = exposure.rescale_intensity(img, in_range=(p3, p80))

plt.figure(figsize=(10, 6))

plt.subplot(2,1,1)
plt.imshow(img_rescale, cmap='gray')
plt.title('Rescaled Image (3% - 80%)')
plt.axis('off')

plt.subplot(2,1,2)
plt.hist(img_rescale.flat, bins=256, range=(0,255))
plt.title('Histogram')
plt.xlabel('Pixel Intensity')
plt.ylabel('Frequency')

plt.tight_layout()
plt.show()
from skimage import exposure

img_equalized = exposure.equalize_hist(img)

plt.figure(figsize=(10, 6))

plt.subplot(2,1,1)
plt.imshow(img_equalized, cmap='gray')
plt.title('Histogram Equalized Image')
plt.axis('off')

plt.subplot(2,1,2)
plt.hist(img_equalized.flat, bins=256, range=(0,1))
plt.title('Equalized Histogram')
plt.xlabel('Pixel Intensity')
plt.ylabel('Frequency')

plt.tight_layout()
plt.show()
from skimage import data
from skimage.exposure import match_histograms


source = data.rocket()    
reference = data.chelsea()


matched = match_histograms(source, reference, channel_axis=-1)


plt.figure(figsize=(12,6))

plt.subplot(1,3,1)
plt.imshow(source)
plt.title('Source (Rocket)')
plt.axis('off')

plt.subplot(1,3,2)
plt.imshow(reference)
plt.title('Reference (Chelsea)')
plt.axis('off')

plt.subplot(1,3,3)
plt.imshow(matched)
plt.title('Matched Image')
plt.axis('off')

plt.tight_layout()
plt.show()
```
